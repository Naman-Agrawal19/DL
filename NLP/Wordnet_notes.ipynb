## **1. WHAT IS WORDNET?**

**WordNet** is a large lexical database (dictionary) of English, developed at Princeton University. It groups English words into sets of synonyms called **synsets**, and records various semantic and lexical relationships between them.


| **🎯  Think of it this way:** |
| - |
| A normal dictionary tells you the DEFINITION of a word. |
| WordNet tells you HOW words RELATE to each other. |
| Example: It knows that 'dog' is a type of 'animal', that 'poodle' is a type of 'dog', |
| and that 'dog' is the opposite of 'cat' in some contexts. |


### **1.1) Why is WordNet Important for NLP?**

WordNet is one of the most widely used resources in Natural Language Processing. It helps computers understand:

- **Word Meaning:** 'bank' as a financial institution vs. 'bank' as a riverbank
- **Word Relationships:** 'car' is similar to 'vehicle', 'automobile', 'motorcar'
- **Semantic Similarity:** How similar are the words 'doctor' and 'nurse'?
- **Text Classification:** Grouping news articles by topic
- **Question Answering:** Understanding that 'canine' = 'dog'

### **1.2) WordNet as an NLTK Corpus**

In Python, WordNet is accessed through the NLTK (Natural Language Toolkit) library. It is treated as a corpus reader — a structured data source that NLTK can query.

Two ways to import WordNet:

```python
from nltk.corpus import wordnet         \# Full name
from nltk.corpus import wordnet as wn   \# Short alias (recommended)
```

## **3. CORE TERMINOLOGY — DEEP DIVE**

Before writing any code, you must understand these core terms. Every WordNet operation revolves around them.
| **TERM** | **MEANING** |
| :-: | - |
| **WORD** | A regular English word like 'dog', 'run', 'happy'. A single word can have multiple meanings. |
| **SENSE** | One specific meaning of a word. 'Bank' has at least 2 senses: <br>(1) financial institution, <br>(2) riverbank. |
| **SYNSET** | A Synonym Set — a group of words that share the same meaning. E.g., \{car, auto, automobile, motorcar\} is one synset. |
| **LEMMA** | A specific word inside a specific synset. It represents one word-sense pairing. E.g., Lemma('dog.n.01.domestic\_dog') |
| **POS** | Part of Speech — whether a word is a Noun (`n`), Verb (`v`), Adjective (`a/s`), or Adverb (`r`). |
| **HYPERNYM** | The '`is-a`' parent. Animal is a hypernym of Dog **\`Dog is a animal\`**. |
| **HYPONYM** | The '`is-a`' child. Dog is a hyponym of Animal **\`Dog is a animal\`.** |
| **HOLONYM** | The '`part-of`' relationship. 'pack' is a holonym of 'dog' **\`dogs are part of a pack\`**. |
| **MERONYM** | The '`has-part`' relationship. 'dog' is a meronym of 'pack'. |
| **ANTONYM** | The opposite word. 'good' and 'bad' are antonyms. |



### **3.1) The Synset ID Format**

Every synset has a 3-part identifier: word.pos.nn


| **Part** | **Meaning** | **Example** |
| :-: | :-: | :-: |
| word | The base word | dog |
| pos | n=noun, v=verb, a=adj, r=adverb | n |
| nn | Sense number (01 = most common) | 01 |


So dog.n.01 means: the word 'dog', as a noun, first/most common sense.

*Example:*



````python
for i in wn.synsets('quote'):
    print(f"{str(i)}, \nDef: {str(i.definition())}, \nExample: {i.examples()}\nPOS: {i.pos()}\n------")
````

## **5. LEMMAS — WORD FORMS**

**A Lemma** in WordNet is a specific word paired with a specific synset (meaning). While a Synset represents a concept, a Lemma represents a WORD that expresses that concept.


| **Synset vs Lemma — The Key Difference** |
| - |
| SYNSET = the CONCEPT  (e.g., the idea of a domesticated dog) |
| LEMMA  = a WORD for that concept  (e.g., 'dog', 'domestic\_dog', 'Canis\_familiaris') |
| One synset can have MULTIPLE lemmas (synonyms). |
| One lemma belongs to EXACTLY ONE synset. |

### **5.1) Accessing Lemmas**


````python
from nltk.corpus import wordnet as wn
# Get lemma directly using word.pos.sensenum.lemma_name format

lemma = wn.lemma('dog.n.01.dog')
print(lemma)           # Lemma('dog.n.01.dog')
print("Name:", lemma.name())  # dog
print("Synset:", lemma.synset())  # Synset('dog.n.01')
print("Key:", lemma.key())     # Unique sense key string
print("Count:", lemma.count())   # How often used in SemCor corpus
````

### **5.2) Lemma Relationships**

Some relations in WordNet only exist at the LEMMA level, not the synset level:

````python
# ANTONYMS — opposite words (only on lemmas!)
good = wn.lemma('good.a.01.good')
print("Antonyms:", good.antonyms()[0].name())

# DERIVATIONALLY RELATED FORMS
vocal = wn.lemma('vocal.a.01.vocal')
print("Derivationally Related Forms:", vocal.derivationally_related_forms()[0])

# PERTAINYMS — adjective derived from a noun
print("Pertainyms:", vocal.pertainyms()[0])
````

| **Relation** | **Description** | **Example** |
| - | - | - |
| `antonyms()` | Opposite in meaning | good → bad |
| `derivationally_related_forms()` | Different POS from same root | vocal → vocalize |
| `pertainyms()` | Adj derived from a noun | vocal → voice |

## **6. LEXICAL RELATIONS**


WordNet organizes words in a rich hierarchy of relationships. Understanding these is critical for NLP tasks.


### **6.1) The WordNet Hierarchy**

The most important structure is the **IS-A hierarchy** (also called the hypernym/hyponym tree). It works exactly like a taxonomy in biology:

### **6.2) Hypernyms — Going UP the Tree**
**Hypernym** = the more GENERAL category. 'Dog' is a type of 'canine'. 'Canine' is the hypernym of 'Dog'.

````python
dog = wn.synset('dog.n.01')

# Immediate parent(s)
print("Hypernyms of 'dog':", dog.hypernyms())   

# All ancestors up to root
hyper = lambda s: s.hypernyms()
ancestors = list(dog.closure(hyper))
print("Ancestors of 'dog':", sorted(ancestors))

# Get the very top (root) of the tree
print("Root hypernyms of 'dog':", dog.root_hypernyms())
````

### **6.3) Hyponyms — Going DOWN the Tree**
**Hyponym** = a more SPECIFIC category. 'Poodle' is a type of 'dog'. 'Poodle' is a hyponym of 'Dog'.

````python
# Immediate children
display("Direct hyponyms (children): (top 5)", sorted(dog.hyponyms())[:5])  # First 5 only

# All descendants (transitive closure)
hypo = lambda s: s.hyponyms()
all_dogs = sorted(dog.closure(hypo))[:5]
display("All descendants:", all_dogs)
````

### **6.4) Finding the Common Ancestor**

**Lowest Common Hypernym (LCH)** = the most specific concept that two words share as an ancestor. This is used to measure semantic similarity.

````python
dog = wn.synset('dog.n.01')
cat = wn.synset('cat.n.01')
woman = wn.synset('woman.n.01')
man = wn.synset('man.n.01')

# What do dog and cat have in common?
print("Lowest common hypernym of 'dog' and 'cat':", dog.lowest_common_hypernyms(cat)[0])

# What do dog and woman have in common?
print("Lowest common hypernym of 'dog' and 'woman':", man.lowest_common_hypernyms(woman)[0])
````

### **6.5) Holonyms and Meronyms — Part/Whole Relations**
**Holonym** is the entire object (e.g., "car"), while a **meronym** is a constituent part of that object (e.g., "wheel").

````python
dog = wn.synset('dog.n.01')

# MEMBER holonyms: 'dog' is a member of what?
print("member holonyms of dog are: ", dog.member_holonyms())   # dog is a member of a 'pack'

# PART meronyms: what are the parts of 'dog'?
body = wn.synset('body.n.01')
print("meronym of body is: ", body.part_meronyms()[:3])   # body has parts: arm, leg, head...

# SUBSTANCE meronyms: made of what?
wood = wn.synset('wood.n.01')
print("substance meronyms of wood are: ", wood.substance_meronyms())
````

| **Method** | **Meaning** | **Example** |
| - | - | - |
| hypernyms() | More general (parent) | dog → canine → animal |
| hyponyms() | More specific (child) | dog → poodle, corgi |
| root\_hypernyms() | Top of the hierarchy | dog → entity.n.01 |
| lowest\_common\_hypernyms() | Shared ancestor | dog+cat → carnivore |
| member\_holonyms() | X is member of what? | dog → pack |
| member\_meronyms() | What are members of X? | pack → dog |
| part\_holonyms() | X is part of what? | finger → hand |
| part\_meronyms() | What are parts of X? | hand → finger, palm |
| substance\_holonyms() | X is substance of what? | H2O → water |
| substance\_meronyms() | Made of what substance? | wood → lignin |
| closure(relation) | All transitive relations | all ancestors of dog |

## **7. SIMILARITY METRICS**
WordNet can calculate how **semantically similar** two words are. This is extremely useful for NLP tasks like finding synonyms, topic modeling, and building chatbots.


| **Setup for all similarity examples:** |
| - |
| from nltk.corpus import wordnet as wn |
| from nltk.corpus import wordnet\_ic |
| dog = wn.synset('dog.n.01') |
| cat = wn.synset('cat.n.01') |
| hit = wn.synset('hit.v.01') |
| slap = wn.synset('slap.v.01') |
| brown\_ic = wordnet\_ic.ic('ic-brown.dat')   \# Information Content corpus |
| semcor\_ic = wordnet\_ic.ic('ic-semcor.dat') \# Another IC corpus |


### **7.1) Path Similarity**
**`path_similarity()`** counts how many steps through the hypernym tree separate two words. `Shorter path = more similar`.

*Formula:* 
$$\text{score} = \frac{1}{\text{shortest\_path\_length} + 1}$$
→  Range: $0$ (unrelated) to $1$ (identical)

````python
dog = wn.synset('dog.n.01')
cat = wn.synset('cat.n.01')
hit = wn.synset('hit.v.01')
slap = wn.synset('slap.v.01')

print("Path similarity between dog and cat:", dog.path_similarity(cat))     # Similar — both mammals
print("Path similarity between hit and slap:", round(hit.path_similarity(slap),3))    # Less similar — both actions but different

# You can also call it on the wn module directly:
print("Path similarity between hit and slap (direct call):", round(wn.path_similarity(hit, slap),3))

# Disable fake root node (may return None for verbs):
print("Path similarity between hit and slap (no fake root):", hit.path_similarity(slap, simulate_root=False))
````

### **6.2) Leacock-Chodorow (LCH) Similarity**

**lch\_similarity()** improves on path similarity by accounting for the DEPTH of the taxonomy. A path in a deep taxonomy is more meaningful than the same path in a shallow one.

*Formula:* 
$$\text{score} = -\ln\big(\frac{\text{path\_length}}{2 \times \text{taxonomy\_depth}}\big)$$

````python
print(round(dog.lch_similarity(cat),3))   # Higher = more similar
print(round(hit.lch_similarity(slap),3))
````

### **6.3) Wu-Palmer (WUP) Similarity**

**wup_similarity()** uses the DEPTH of both synsets and their Least Common Subsumer (LCS) in the hierarchy. Very popular in practice.

*Formula:* 
$$\text{score} = \frac{2 \times \text{depth(LCS)}}{\text{depth(s1)} + \text{depth(s2)}}$$


````python
print(round(hit.wup_similarity(slap), 3))
print(round(dog.wup_similarity(cat), 3))
````

### **6.4) Information Content (IC) Based Metrics**

These metrics use **Information Content (IC)** — a measure of how rare/specific a word is in a large corpus. Rare, specific words carry more information. You need to load an IC file first.

#### **6.4.1) Resnik Similarity**

Based only on the IC of the LCS (common ancestor). 
$$\text{score} = \text{IC}(\text{LCS})$$

````python
from nltk.corpus import wordnet_ic
brown_ic = wordnet_ic.ic('ic-brown.dat')
semcor_ic = wordnet_ic.ic('ic-semcor.dat')

print(round(dog.res_similarity(cat, brown_ic), 3))  
print(round(dog.res_similarity(cat, semcor_ic), 3))  # Different corpus = different score!
````


#### **6.4.2) Jiang-Conrath (JCN) Similarity**

Considers IC of both words AND their LCS. 

$$\text{score} = \frac{1}{(\text{IC(s1)} + \text{IC(s2)} - 2 \times \text{IC(LCS)})}$$

````python
print(round(dog.jcn_similarity(cat, brown_ic), 3))
````

#### **6.4.3) Lin Similarity**

Normalized version of Resnik. 

$$\text{score} = \frac{2 \times \text{IC(LCS)}}{(\text{IC(s1)} + \text{IC(s2)})}$$
→  Range: 0 to 1

````python
print(round(dog.lin_similarity(cat, semcor_ic),3))
````

89% similar — dog and cat are very related concepts!

#### **Comparison of All Similarity Metrics**

| **Metric** | **Based On** | **Range** | **Needs IC?** | **dog vs cat** |
| - | - | - | - | - |
| path\_similarity | Shortest path | 0 to 1 | No | 0.2 |
| lch\_similarity | Path + depth | \>0 (log scale) | No | 2.028 |
| wup\_similarity | Depths of synsets | 0 to 1 | No | 0.857 |
| res\_similarity | IC of LCS | 0 to ∞ | Yes | 7.911 |
| jcn\_similarity | IC of both + LCS | 0 to ∞ | Yes | 0.449 |
| lin\_similarity | Normalized IC | 0 to 1 | Yes | 0.886 |

## **8. MORPHY — LEMMATIZATION**

**Morphy** is WordNet's built-in lemmatizer. It converts inflected word forms back to their base (lemma) form. For example, 'dogs' → 'dog', 'running' → 'run', 'better' → 'good'.


| **Why do we need Morphy?** |
| - |
| WordNet stores only BASE FORMS. So if you search for 'dogs', you won't find it directly. |
| Morphy converts 'dogs' → 'dog', then WordNet can find the synsets. |
| This is called LEMMATIZATION — a key NLP preprocessing step. |


### **8.1) Using wn.morphy()**


````python
from nltk.corpus import wordnet as wn

# Basic lemmatization
print(wn.morphy('dogs'))         # plural noun → base
print(wn.morphy('churches'))     # irregular plural
print(wn.morphy('aardwolves'))   # very irregular plural
print(wn.morphy('abaci'))        # Latin plural → abacus


# With POS context
print(wn.morphy('denied', wn.VERB))   # verb inflection
print(wn.morphy('denied', wn.NOUN))   # None — 'denied' is not a noun

# Now search using the lemmatized form
display(wn.synsets('denied', wn.VERB))  # WordNet auto-lemmatizes
````

### **8.2 WordNetLemmatizer**

NLTK also provides a WordNetLemmatizer class that wraps Morphy for easy use:

````python
from nltk.stem.wordnet import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
print("eggs =>",lemmatizer.lemmatize('eggs', pos='n'))   # noun
print("legs =>",lemmatizer.lemmatize('legs', pos='n'))   # noun
print("running =>",lemmatizer.lemmatize('running', pos='v')) # verb
print("better =>",lemmatizer.lemmatize('better', pos='a'))  # adjective
````

### **Use Case 1: Word Sense Disambiguation (WSD)**

Problem: The word 'bank' can mean a financial institution or the side of a river. **WSD** picks the right sense based on context.

````python
from nltk.corpus import wordnet as wn
def get_definition(word, sense_number):
    synsets = wn.synsets(word, pos=wn.NOUN)
    if sense_number <= len(synsets):
        syn = synsets[sense_number - 1]
        return {
            'synset': syn.name(),
            'definition': syn.definition(),
            'examples': syn.examples()
        }
    return None
````

````python
word = 'bank'
for i in range(1, 4):
    result = get_definition(word, i)
    if result:
        print(f'Sense {i}: {result["definition"]}')
````

### **Use Case 2: Semantic Similarity Between Sentences**

Compare two sentences by finding the maximum similarity between any word pair.

````python
from nltk.corpus import wordnet as wn
import nltk
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')

def get_first_synset(word, pos='n'):
    synsets = wn.synsets(word, pos=pos)
    return synsets[0] if synsets else None

def sentence_similarity(sent1_words, sent2_words):
    """Compare each word in sent1 with each word in sent2."""
    scores = []
    for w1 in sent1_words:
        s1 = get_first_synset(w1)
        if not s1: continue
        for w2 in sent2_words:
            s2 = get_first_synset(w2)
            if not s2: continue
            sim = s1.wup_similarity(s2)
            if sim is not None:
                scores.append(sim)
    return sum(scores)/len(scores) if scores else 0
````

````python
# Example

s1 = ['dog', 'run', 'park']
s2 = ['puppy', 'jog', 'garden']
s3 = ['car', 'drive', 'highway']
print(f'S1 vs S2: {sentence_similarity(s1, s2):.3f}')  # Should be high
print(f'S1 vs S3: {sentence_similarity(s1, s3):.3f}')  # Should be lower
````

### **Use Case 3: Hypernym Path Explorer**

Useful for taxonomy-based text classification — trace a word up to its root.

````python
def hypernym_path(word, pos='n'):
    """Show the full IS-A path for a word up to entity."""
    synsets = wn.synsets(word, pos=pos)
    if not synsets:
        return f'No synsets found for {word}'
    
    syn = synsets[0]  # Take most common sense
    paths = syn.hypernym_paths()
    path = paths[0]  # Take first path    

    print(f'Hypernym path for "{word}" ({syn.name()}):',)

    for i, s in enumerate(path):
        indent = '  ' * i
        print(f'{indent}→ {s.name()} ({s.definition()[:40]}...)')
````

````python
hypernym_path('poodle')
````

### **Use Case 4: Synonym Expansion for Search**
Improve search recall by expanding query words with their synonyms.

````python
def expand_query(query_word, pos='n', max_synonyms=5):
    """Expand a search term with its WordNet synonyms."""
    synonyms = set()
    synonyms.add(query_word)  # include original
    for synset in wn.synsets(query_word, pos=pos):
        for lemma in synset.lemmas():
            name = lemma.name().replace('_', ' ')
            synonyms.add(name)   
    return list(synonyms)[:max_synonyms + 1]

print('Synonyms for car:', expand_query('car'))
print('Synonyms for happy:', expand_query('happy', pos='a'))
print('Synonyms for run:', expand_query('run', pos='v'))
````

### **Use Case 5: Checking if Word Belongs to a Category**

Check if a word is a type of a given category using hypernym chains.


````python
def is_a(word, category, pos='n'):
    """Check if 'word' is a type of 'category' using WordNet."""
    word_synsets = wn.synsets(word, pos=pos)
    cat_synsets = wn.synsets(category, pos=pos)
    if not word_synsets or not cat_synsets:
        return False

    word_syn = word_synsets[0]
    cat_syn = cat_synsets[0]

    # Get all hypernyms (ancestors) of the word
    hyper = lambda s: s.hypernyms()
    all_hypernyms = set(word_syn.closure(hyper))
    return cat_syn in all_hypernyms
````

````python
# Test it out
print(is_a('poodle', 'dog'))     # True — poodle IS a dog
print(is_a('poodle', 'animal'))  # True — poodle IS an animal
print(is_a('car', 'animal'))     # False — car is NOT an animal
print(is_a('eagle', 'bird'))     # True — eagle IS a bird
````
