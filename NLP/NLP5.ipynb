```python
from gensim.models import Word2Vec

corpus = [
    ["the", "cat", "sat", "on", "the", "mat"],
    ["the", "dog", "played", "in", "the", "park"],
    ["the", "cat", "and", "dog", "are", "friends"],
    ["cats", "and", "dogs", "play", "together"],
    ["the", "mat", "is", "on", "the", "floor"],
]

# sg=0 → CBOW
model_cbow = Word2Vec(sentences=corpus, vector_size=15,
                      window=2, min_count=1, sg=0, epochs=100)
print("Vector for 'cat':")
display( model_cbow.wv['cat'].round(3))
print("Most similar to 'cat':")
display(model_cbow.wv.most_similar('cat', topn=3))
print("Most similar to 'dog':")
display(model_cbow.wv.most_similar('dog', topn=3))
```

```python
from gensim.models import Word2Vec

corpus = [
    ["the", "cat", "sat", "on", "the", "mat"],
    ["the", "dog", "played", "in", "the", "park"],
    ["the", "cat", "and", "dog", "are", "friends"],
    ["cats", "and", "dogs", "play", "together"],
    ["the", "mat", "is", "on", "the", "floor"],
]

# sg=1 → Skip-gram
model_skipgram = Word2Vec(sentences=corpus, vector_size=15,
                          window=2, min_count=1, sg=1, epochs=100)

print("Vector for 'cat':")
display( model_skipgram.wv['cat'].round(3))
print("Most similar to 'cat':")

display(model_skipgram.wv.most_similar('cat', topn=3))
print("Most similar to 'dog':")
display(model_skipgram.wv.most_similar('dog', topn=3))
```
