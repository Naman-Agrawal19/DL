````python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# Define the model
model = Sequential([
    Input(shape=(2,), name="Input_Layer"),  # Input layer with 2 neurons
    Dense(2, activation='relu', name="Hidden_Layer"),  # Hidden layer with 2 neurons
    Dense(1, activation='linear', name="Output_Layer")  # Output layer with 1 neuron
])

# Compile the model (dummy compilation for visualization purposes)
model.compile(optimizer='adam', loss='binary_crossentropy')

# Summary of the model
model.summary()

# Setup TensorBoard logging
log_dir = "logs/network_visualization"
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir=log_dir, histogram_freq=1)

# Dummy data (to create computation graph)
import numpy as np
x_dummy = np.random.random((10, 2))  # 10 samples, 2 features
y_dummy = np.random.randint(0, 2, (10, 1))  # 10 samples, binary labels

# Train the model (to generate logs for TensorBoard)
model.fit(x_dummy, y_dummy, epochs=1, callbacks=[tensorboard_callback])

# To visualize:
# Run the following command in your terminal:
%tensorboard --logdir logs/network_visualization
````

````python
import graphviz

def draw_neural_network_graphviz(layer_sizes, weights=None, filename="neural_network"):
    """
    Draws a neural network diagram using Graphviz.

    Args:
        layer_sizes: A list of integers representing the number of neurons in each layer.
        weights: A list of weight matrices (optional).
        filename: The name of the output file (without extension).
    """

    dot = graphviz.Digraph(comment='Neural Network', graph_attr={'rankdir': 'LR'})  # Left-to-right layout

    for i, layer_size in enumerate(layer_sizes):
        with dot.subgraph(name=f'cluster_{i}') as c:
            c.attr(label=f'Layer {i+1}', style='rounded', rank='same') #Added layer name and style
            for j in range(layer_size):
                node_name = f'l{i}_n{j}'
                if i == 0:
                    label = f'x{j+1}'
                elif i == len(layer_sizes) - 1:
                    label = 'o'
                else:
                    label = f'h{j+1}'
                c.node(node_name, label=label, shape='circle')

    if weights is not None:
        for i in range(len(layer_sizes) - 1):
            for j in range(layer_sizes[i]):
                for k in range(layer_sizes[i+1]):
                    source = f'l{i}_n{j}'
                    target = f'l{i+1}_n{k}'
                    weight_val = weights[i][j, k]
                    dot.edge(source, target, label=f'{weight_val:.2f}') #weight with 2 decimal places

    dot.render(filename, view=True)  # Saves the graph to a file and opens it

# Example usage: 2 inputs, 2 hidden, 1 output (without weights)
layer_sizes = [2, 2, 1]
draw_neural_network_graphviz(layer_sizes)

# Example with weights:
import numpy as np
layer_sizes = [2, 2, 1]
weights = [np.random.rand(2, 2), np.random.rand(2, 1)]
draw_neural_network_graphviz(layer_sizes, weights, filename="neural_network_with_weights")
````
